```python
from bs4 import BeautifulSoup
import smtplib
import requests
import time
import datetime
import csv
import pandas as pd
```

```python
url = 'https://www.amazon.in/dp/B0H1X5XP25?pf_rd_r=4KMZEVCKPJFBEAYXHBG2&pf_rd_p=3b524752-05d9-4526-b2f6-90828b2c03f8&pf_rd_m=A1VBAL9TL5WCBF&pf_rd_s=merchandised-search-6&pf_rd_t&pf_rd_i=1389401031'
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/78.0.3904.108 Safari/537.36", "Accept-Encoding":"gzip, deflate", "Accept":"text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8", "DNT":"1","Connection":"close", "Upgrade-Insecure-Requests":"1"}
header = ['Title','Price','Date']

with open('Samsung_A56_Price.csv','w',newline="") as f:
        write = csv.writer(f)
        write.writerow(header)



def check_price():
    page = requests.get(url,headers=headers)
    s = BeautifulSoup(page.content,'html.parser')
    sp = BeautifulSoup(s.prettify(),'html.parser')
    title = sp.find(id="productTitle").get_text(strip=True)
    price = sp.find(class_="a-price-whole").get_text(strip=True)
    date = datetime.date.today()
    #print(title)
    #print(price)
    #print(date)
    import csv
    import pandas as pd

    data = [title,price,date]

    with open('Samsung_A56_Price.csv','a+',newline="") as f:
        write = csv.writer(f)
        write.writerow(data)

```

```python
while(True):
    check_price()
    time.sleep(86400)
```
